```python
import tensorflow as tf
from tensorflow import keras
```

The rule for ReLU is incredibly simple: If the input is negative, change it to zero. If the input is positive, leave it exactly as it is.

FLATTEN is used to cinvert the multi dimentional into single dimention


```python
model=keras.Sequential([
    keras.layers.Flatten(input_shape=(30,)),#30 is columns
    #the number 32 is based 2 power n ex:(2poww3)=8 because we have columns 30 --> near to 30 is 2 power 5 that is 32
    keras.layers.Dense(32,activation='relu'),#-->relu(Rectified Linear Unit) is used to convert the negative into positive
    keras.layers.Dense(2,activation='sigmoid')
])
```

The optimizer is the algorithm responsible for updating the weights of your neural network to reduce the error.∇

The loss function measures how wrong the model's predictions are. It calculates a single penalty score for every mistake the model makes; the goal of the optimizer is to get this score as close to zero as possible.

```python
model.compile(optimizer='adam',
loss='sparse_categorical_crossentropy',
             metrics=(['accuracy'])
)
```

```python
history=model.fit(x_train_s,y_train,validation_split=0.1,epochs=50)
```

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title('model accuracy')
plt.xlabel('epoch')
plt.ylabel('accuracy')
plt.legend(['training data0','validation data'],loc='lower right')
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('model accuracy')
plt.xlabel('epoch')
plt.ylabel('accuracy')
plt.legend(['training data0','validation data'],loc='lower right')
```
