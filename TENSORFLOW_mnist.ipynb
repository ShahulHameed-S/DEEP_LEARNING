```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn.datasets
from sklearn.model_selection import train_test_split
```

```python
from tensorflow import keras as kr
from tensorflow.keras import layers
```

```python
d=kr.datasets.mnist
(x_train,y_train),(x_test,y_test)=d.load_data()
x_train,x_test=x_train/255.0,x_test/255.0
#when we do this it will become standardized so that it range conveerts into 0 to 1 which is a gray scale
```

Softmax is used to chanelize the output for  ex op is 9 then it defines wether the op is 9 or 1 or any other numbers


```python
model=kr.Sequential([
    kr.layers.Flatten(input_shape=(28,28,1)),
    kr.layers.Dense(128,activation="relu"),
    kr.layers.Dense(10,activation="softmax")#10 is due to there are 10 outputs in y_train label(0 to 9)
])
```

```python
model.compile(optimizer='adam',
loss='sparse_categorical_crossentropy',
             metrics=(['accuracy'])
)
```

```python
model=model.fit(x_train,y_train,validation_split=0.1,epochs=10)
```

```python
plt.plot(model.history['loss'])
plt.plot(model.history['val_loss'])
plt.title('model accuracy')
plt.xlabel('epoch')
plt.ylabel('accuracy')
plt.legend(['training data0','validation data'],loc='lower right')
```
