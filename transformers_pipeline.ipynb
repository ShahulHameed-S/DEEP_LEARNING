```python
from transformers import pipeline

qa = pipeline("question-answering", model="deepset/roberta-base-squad2")

result = qa(
    question="Where is he standing?",
    context="A man is standing on the chair in the room."
)

print(result)
```

```python
from transformers import pipeline

summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

text = """
Artificial intelligence is transforming the world. It is used in healthcare,
education, transportation, and many other fields. AI helps automate tasks
and improve decision making.
"""

summary = summarizer(text)
print(summary)
```

```python
from transformers import pipeline

fill = pipeline("fill-mask", model="roberta-base")

result = fill("I am standing on the <mask>.")

print(result)
```

```python
from transformers import pipeline

translator = pipeline(
    "translation",
    model="Helsinki-NLP/opus-mt-en-mul"
)

result = translator("Hello, how are you?")
print(result)
```
